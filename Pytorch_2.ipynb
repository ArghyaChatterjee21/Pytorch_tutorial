```python
b= torch.rand(2,3,4)
print(b)
```

```python
#Unsqueeze
#image size
c = torch.rand(226,226,3)
print(c.shape)
print(c)

print(c.unsqueeze(0).shape)
```

```python
print(c.shape)
print(c.unsqueeze(0).shape)
print(c.unsqueeze(1).shape)
print(c.unsqueeze(2).shape)
print(c.unsqueeze(3).shape)
```

```python
#squeeze
d = torch.rand(1,20)
print(d.shape)
print(d)
print(d.squeeze(0).shape)
print(d.squeeze(0))
```
